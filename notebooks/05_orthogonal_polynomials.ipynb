```python
#Orthogonal Polynomial Approximation

import numpy as np
import matplotlib.pyplot as plt


def f(x):
    return np.sin(np.pi*x)

def trZ(a,b,n,f):

    x=np.linspace(a,b,n)

    h=(b-a)/(n)
    k=0
    for i in range(1,n):
        k=k+f(x[i])

    tr=(h/2)*(f(x[0])+2*k+f(x[-1]))
    return tr

a=0
b=1
n=100
k=3
basis1=np.array([lambda x:1,lambda x:x,lambda x:x**2])
basis2=np.array([lambda x:x,lambda x:x**2,lambda x:x**3])
basis3=np.array([lambda x:x**2,lambda x:x**3,lambda x:x**4])
A=np.zeros((k,k))

for i,gi in enumerate(basis1):
    A[0][i]=trZ(a,b,n,gi)
for i,gi in enumerate(basis2):
    A[1][i]=trZ(a,b,n,gi)
for i,gi in enumerate(basis3):
    A[2][i]=trZ(a,b,n,gi)


y=np.zeros(k)
Y=np.array([lambda x:np.sin(np.pi*x),lambda x:x*np.sin(np.pi*x),lambda x:x**(2)*np.sin(np.pi*x)])
for i,yi in enumerate(Y):
    y[i]=trZ(a,b,n,yi)

c=np.linalg.solve(A,y)
print(c)



```

```python
#A more accurate with sympy integration Orthogonal Polynomial Approximation

import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

x = sp.symbols('x')

def f(x):
    return np.sin(np.pi*x)

def sympy_integrate(func, a, b):
    """Integrates a function using sympy."""
    x = sp.symbols('x')
    func_sp = sp.sympify(func(x))  # Convert the lambda function to a sympy expression
    result = sp.integrate(func_sp, (x, a, b))
    return result.evalf()  # Evaluate the result to a numerical value


a = 0
b = 1
n = 100
k = 3

basis1 = np.array([lambda x: 1, lambda x: x, lambda x: x**2])
basis2 = np.array([lambda x: x, lambda x: x**2, lambda x: x**3])
basis3 = np.array([lambda x: x**2, lambda x: x**3, lambda x: x**4])

A = np.zeros((k, k))

for i, gi in enumerate(basis1):
    A[0][i] = sympy_integrate(gi, a, b)
for i, gi in enumerate(basis2):
    A[1][i] = sympy_integrate(gi, a, b)
for i, gi in enumerate(basis3):
    A[2][i] = sympy_integrate(gi, a, b)

y = np.zeros(k)
Y = np.array([lambda x: sp.sin(sp.pi*x), lambda x: x*sp.sin(sp.pi*x), lambda x: x**(2)*sp.sin(sp.pi*x)])

for i, yi in enumerate(Y):
    y[i] = sympy_integrate(yi, a, b)

c = np.linalg.solve(A, y)
print(c)
```

```python
#Generalized Orthogonal Functions Finding by weight function


import sympy as sp
import matplotlib.pyplot as plt

x = sp.symbols('x')

def pol(a,b,w,n):
    P = [0]*n
    A = [0]*n
    B = [0]*n
    P[0] = 1
    P[1] = x - sp.integrate(x*w*(P[0]**2), (x, a, b))/sp.integrate(w*(P[0]**2), (x, a, b))
    for i in range(2, n):
        A[i] = sp.integrate(x * w * P[i-1]**2, (x, a, b)) / sp.integrate(w*P[i-1]**2, (x, a, b))
        B[i] = sp.integrate(x * w * P[i-1] * P[i-2], (x, a, b)) / sp.integrate(w*P[i-2]**2, (x, a, b))
        P[i] = (x - A[i]) * P[i-1] - B[i] * P[i-2]

    return P

n = 6
w=sp.exp(-x**2)
a = -sp.oo
b = sp.oo

w=1
a=-1
b=1

P = pol(a,b,w,n)

print(P)


X=np.linspace(a,b,1000)
Y=np.zeros((len(P),len(X)))
for i in range(len(P)):
  numpy_func = sp.lambdify(x, P[i], 'numpy')
  Y[i] = numpy_func(X)
  if i>0:
    plt.plot(X,Y[i])
    plt.grid()
```
