```python
import numpy as np

def newton_multivariate_manual_jac(functions, jacobian, x0, tol=1e-6, max_iter=100):
    """
    Finds roots of a system of nonlinear equations using Newton's method with manual Jacobian.
    Handles singular Jacobian matrices gracefully.
    """
    x = np.array(x0, dtype=float)
    for _ in range(max_iter):
        f = np.array([func(x) for func in functions], dtype=float)

        # Check convergence
        if np.linalg.norm(f) < tol:
            return x

        # Compute Jacobian
        n = len(x)
        J = np.zeros((n, n))
        for i in range(n):
            for j in range(n):
                J[i, j] = jacobian[i][j](x)

        # Solve using least-squares to handle ill-conditioned matrices
        try:
            delta_x = np.linalg.lstsq(J, f, rcond=None)[0]
        except np.linalg.LinAlgError:
            print("Warning: Ill-conditioned Jacobian. Trying perturbed step.")
            delta_x = np.zeros_like(x)

        # Check for invalid steps
        if np.any(np.isnan(delta_x)):
            print("Warning: NaN in step. Resetting perturbation.")
            delta_x = 0.1 * np.ones_like(x)

        x -= delta_x

        if np.linalg.norm(delta_x) < tol:
            return x

    print("Warning: Max iterations reached.")
    return x

# Constants and system setup (same as before)
V_plus = 5.0
R1, R2, R3, R4 = 1000.0, 4000.0, 3000.0, 2000.0
J0, VT = 3e-9, 0.05

def f1(x):
    V1, V2 = x
    return (V1 - V_plus)/R1 + V1/R2 + J0*(np.exp((V1-V2)/VT) - 1)

def f2(x):
    V1, V2 = x
    return J0*(np.exp((V1-V2)/VT) - 1) - (V2/R3 + V2/R4)

jacobian = [
    [lambda x: (1/R1 + 1/R2) + (J0/VT)*np.exp((x[0]-x[1])/VT),
     lambda x: -(J0/VT)*np.exp((x[0]-x[1])/VT)],
    [lambda x: (J0/VT)*np.exp((x[0]-x[1])/VT),
     lambda x: -(J0/VT)*np.exp((x[0]-x[1])/VT) - (1/R3 + 1/R4)]
]

# Improved initial guess based on circuit intuition
x0 = [V_plus * 4, V_plus * 3]  # [4.0, 1.5]

# Solve
solution = newton_multivariate_manual_jac([f1, f2], jacobian, x0)
print(f"Converged solution: V1 = {solution[0]:.6f} V, V2 = {solution[1]:.6f} V")


#for the (c) part we use the formula Vd=V1-V2 which in turn is 0.6V so it proves the result
```
