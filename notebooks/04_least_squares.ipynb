```python
#Linear Least Square Approximation

import numpy as np
import matplotlib.pyplot as plt


def LLS(x,y):
    A=np.ones((2,2))
    A[0][0]=np.sum(x**2)
    A[0][1]=np.sum(x)
    A[1][0]=np.sum(x)
    A[1][1]=len(x)
    print(A)
    Y=np.zeros(2)
    Y[0]=np.sum(x*y)
    Y[1]=np.sum(y)
    L=np.linalg.solve(A,Y)
    return L

x=np.array([1,1.25,1.5,1.75,2])
y=np.array([5.1,5.79,6.53,7.45,8.46])
L=LLS(x,y)
print(L)
plt.plot(x,y,'o')
X=np.linspace(1,2,1000)
Y=L[1]+L[0]*X

plt.plot(X,Y,'-')
```

```python
#Logarthmic Least Square Approximation

import numpy as np
import matplotlib.pyplot as plt


def LLS(x,y):
    A=np.ones((2,2))
    A[0][0]=np.sum(x**2)
    A[0][1]=np.sum(x)
    A[1][0]=np.sum(x)
    A[1][1]=len(x)
    print(A)
    Y=np.zeros(2)
    Y[0]=np.sum(x*np.log(y))
    Y[1]=np.sum(np.log(y))
    L=np.linalg.solve(A,Y)
    L[1]=np.exp(L[1])
    return L

x=np.array([1,1.25,1.5,1.75,2])
y=np.array([5.1,5.79,6.53,7.45,8.46])
L=LLS(x,y)
print(L)
plt.plot(x,y,'o')
X=np.linspace(1,2,1000)
Y=L[1]*np.exp(L[0]*X)

plt.plot(X,Y,'-')
```

```python
#Polynomial Square Approximation

import numpy as np
import matplotlib.pyplot as plt

def plot(x,L):
    y=L[0]*x**4+L[1]*x**3+L[2]*x**2+L[3]*x+L[4]
    return y
def PSS(x,y,n):
    n=n
    A=np.zeros((n+1,n+1))
    for i in range(n+1):
        c=np.ones(n+1)
        for j in range(n+1):
            c[j]=np.sum(x**(2*n-i-j))
        A[i]=c
    Y=np.zeros(n+1)
    for j in range(n+1):
        Y[j]=np.sum(y*x**(n-j))
    L=np.linalg.solve(A,Y)
    return L
x=np.array([0,0.25,0.5,0.75,1])
y=np.array([1,1.284,1.6487,2.117,2.7183])


L=PSS(x,y,n=len(x)-1)
print(L)

X=np.linspace(0,1,1000)
Y=plot(X,L)

plt.plot(x,y,'o')
plt.plot(X,Y,'-')

```
