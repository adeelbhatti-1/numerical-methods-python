```python
import numpy as np

import matplotlib.pyplot as plt

a = 1
b = 2
alpha = 1
beta = 2
N = 10
h = (b - a) / N
x = np.linspace(a, b, N + 1)
y = np.zeros(N + 1)
y[0] = alpha
y[N] = beta

def p(x):
    return -2/x

def q(x):
    return 2/x**2

def r(x):
    return np.sin(np.log(x))/x**2

matrix = np.zeros((N, N))

for i in range(N):
    matrix[i, i] = 2 + h**2 * q(x[i + 1])

for i in range(1, N):
    matrix[i, i - 1] = -1 - h / 2 * p(x[i])

for i in range(N - 1):
    matrix[i, i + 1] = -1 + h / 2 * p(x[i + 1])
print(matrix)
rhs = np.zeros(N)
for i in range(N):
    rhs[i] = - h **2 * r(x[i + 1])

rhs[0] = rhs[0] + (1 + h / 2 * p(x[1])) * alpha
rhs[N - 1] = rhs[N - 1] + (1 - h / 2 * p(x[N])) * beta

y = np.linalg.solve(matrix, rhs)
x1 = x[1:]
print(y)
plt.plot(x1, y)
```
