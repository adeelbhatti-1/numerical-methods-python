```python
#Chebyshev Polynomials Generation

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')

def pol(n):
    T = [0] * n
    T[0] = 1
    if n > 1:
        T[1] = x

    for i in range(2, n):
        T[i] = 2 * x * T[i-1] - T[i-2]

    return T

n = 5
P = pol(n)

print(P)

X=np.linspace(-1,1,1000)
Y=np.zeros((len(P),len(X)))
for i in range(len(P)):
  numpy_func = sp.lambdify(x, P[i], 'numpy')
  Y[i] = numpy_func(X)
  if i>0:
    plt.plot(X,Y[i])
    plt.grid()
```

```python
#Minimizing Approximation Error on Arbitrary Intervals

import sympy
import math as m
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

def f(x):
  return x*np.exp(x)

def Lagrange (Lx, Ly):
    X=sympy.symbols('X')
    L=float(0.0)
    for k in range ( len(Lx) ):
        t=float(1.0)
        for j in range ( len(Lx) ):
            if j != k:
                t=t* ( (X-Lx[j]) / float(Lx[k]-Lx[j]) ) # when I put number, OK
        L+= t*Ly[k]
    return L

Lx=np.array([0,0.5,1,1.5])
Ly=f(Lx)
L=Lagrange(Lx,Ly)
xn=np.zeros(len(Lx))
for i in range(len(Lx)):
  xn[i]=np.cos(((2*i+1)/8)*np.pi)

xn=0.5*((Lx[-1]-Lx[0])*xn+Lx[-1]+Lx[0])
yn=f(xn)

Ln=Lagrange(xn,yn)

print('x           f(x)=xe^x        P3(x)       |xe^x-P3(x)|      P^3(x)      |xe^x-P^3(x)|')
print('------------------------------------------------------------------------------------')
xp=np.arange(Lx[0]+0.15,Lx[-1]-0.15,0.15)
for i in xp:
  g1=i
  g2=f(g1)
  g3=L.subs(X,g1).evalf()
  g4=abs(g2-g3) # define g4 before using in string formatting
  g5=Ln.subs(X,g1).evalf()
  g6=abs(g2-g5)
  print('{:.4f}       {:.4f}       {:.4f}        {:.4f}         {:.4f}            {:.4f}'.format(g1,g2,g3,g4,g5,g6))



XE=np.linspace(Lx[0],Lx[-1],1000)

X = sympy.symbols('X')
numpy_func = sp.lambdify(X, L, 'numpy')
numpy_funcn = sp.lambdify(X, Ln, 'numpy')
Y = numpy_func(XE)
Yn=numpy_funcn(XE)

plt.plot(XE,Y)
plt.plot(XE,Yn)
plt.scatter(Lx,Ly)
plt.plot(xn,yn,'o',color='orange')
plt.grid()
plt.show()
```
