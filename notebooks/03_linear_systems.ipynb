```python
from numpy import array, empty

A = array([[ 2,  1,  4,  1 ],
           [ 3,  4, -1, -1 ],
           [ 1, -4,  1,  5 ],
           [ 2, -2,  1,  3 ]], float)

v = array([ -4, 3, 9, 7 ], float)
N = len(v)

# Gaussian elimination
for m in range(N):
    # Divide by the diagonal element
    div = A[m, m]
    A[m, :] /= div
    v[m] /= div

    # Now subtract from the lower rows
    for i in range(m+1, N):
        mult = A[i, m]
        A[i, :] -= mult * A[m, :]
        v[i] -= mult * v[m]

# Backsubstitution
x = empty(N, float)
for m in range(N-1, -1, -1):
    x[m] = v[m]
    for i in range(m+1, N):
        x[m] -= A[m, i] * x[i]

print(x)
```

```python
#included partial pivoting

from numpy import array, empty, argmax

A = array([[ 0,  1,  4,  1 ],
           [ 3,  4, -1, -1 ],
           [ 1, -4,  1,  5 ],
           [ 2, -2,  1,  3 ]], float)

v = array([ -4, 3, 9, 7 ], float)
N = len(v)

# Gaussian elimination with partial pivoting
for m in range(N):
    # Partial pivoting: Find the row with the largest element in the current column
    max_row = argmax(abs(A[m:, m])) + m  # Find the row index with the maximum absolute value in column m
    if max_row != m:
        # Swap rows in A
        A[[m, max_row]] = A[[max_row, m]]
        # Swap corresponding elements in v
        v[m], v[max_row] = v[max_row], v[m]

    # Divide by the diagonal element
    div = A[m, m]
    if div == 0:
        raise ValueError("Matrix is singular or nearly singular.")
    A[m, :] /= div
    v[m] /= div

    # Subtract from the lower rows
    for i in range(m+1, N):
        mult = A[i, m]
        A[i, :] -= mult * A[m, :]
        v[i] -= mult * v[m]

# Backsubstitution
x = empty(N, float)
for m in range(N-1, -1, -1):
    x[m] = v[m]
    for i in range(m+1, N):
        x[m] -= A[m, i] * x[i]

print("Solution vector x:", x)
```

```python
import numpy as np

def LU(A):
  n = len(A)
  L = np.eye(n)  # Initialize L as an identity matrix
  U = A.copy()   # Copy A into U to perform row operations

  for k in range(n - 1):
      for i in range(k + 1, n):
          factor = U[i, k] / U[k, k]  # Compute the multiplier
          L[i, k] = factor  # Store the factor in L
          U[i, k:] -= factor * U[k, k:]  # Update rows in U
  return L,U


```
