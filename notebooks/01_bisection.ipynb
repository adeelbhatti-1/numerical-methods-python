```python
#Example 6.14


import numpy as np
import matplotlib.pyplot as plt

# Constants
m = 9.1094e-31   # Mass of electron (kg)
V = 20 * 1.60218e-19  # Convert eV to Joules
w = 1e-9  # Convert nm to meters
hbar = 6.62607015e-34 / (2 * np.pi)  # Reduced Planck's constant

# Energy range (convert eV to Joules)
E_eV = np.linspace(0, 20, 1000)  # Avoid division by zero
E = E_eV * 1.60218e-19  # Convert to Joules

# Calculate functions
y1 = np.tan(np.sqrt((w**2 * m * E) / (2 * hbar**2)))
y2 = np.sqrt((V - E) / E)
y3 = -np.sqrt(E / (V - E))

# Plotting
plt.figure(figsize=(8, 6))
plt.plot(E_eV, y1, label=r'$y_1 = \tan(\sqrt{w^2 m E / 2\hbar^2})$', color='blue')
plt.plot(E_eV, y2, label=r'$y_2 = \sqrt{(V - E) / E}$', color='red')
plt.plot(E_eV, y3, label=r'$y_3 = -\sqrt{E / (V - E)}$', color='green')

# Labels and legends
plt.axhline(0, color="black", linestyle="--")
plt.xlabel("Energy (eV)")
plt.ylabel("Function Values")
plt.title("Plot of y1, y2, and y3 as a function of Energy")
plt.legend()
plt.ylim(-10, 10)  # Limit y-axis for better visualization
plt.grid()
plt.show()
```

```python
import numpy as np

# Constants
m = 9.1094e-31   # Electron mass (kg)
V = 20 * 1.60218e-19  # Convert eV to Joules
w = 1e-9  # Well width in meters
hbar = 6.62607015e-34 / (2 * np.pi)  # Reduced Planck's constant
tolerance = 0.001 * 1.60218e-19  # Convert eV to Joules

# Function for y1
def f_y1(E):
    return np.tan(np.sqrt((w**2 * m * E) / (2 * hbar**2)))

# Function for y2 (even states)
def f_y2(E):
    return np.sqrt((V - E) / E)

# Function for y3 (odd states)
def f_y3(E):
    return -np.sqrt(E / (V - E))

# Function to find roots using binary search
def binary_search(f1, f2, E_min, E_max):
    while (E_max - E_min) > tolerance:
        E_mid = (E_min + E_max) / 2
        y1 = f1(E_mid)
        y2 = f2(E_mid)

        if np.isnan(y1) or np.isnan(y2):
            return None

        if y1 > y2:
            E_max = E_mid
        else:
            E_min = E_mid
    return (E_min + E_max) / 2

# Finding the first six energy levels
energy_levels = []
for n in range(6):
    E_low = (n * V) / 10  # Initial guess for lower bound
    E_high = ((n + 1) * V) / 10  # Upper bound

    if n % 2 == 0:
        E_root = binary_search(f_y1, f_y2, E_low, E_high)  # Even states
    else:
        E_root = binary_search(f_y1, f_y3, E_low, E_high)  # Odd states

    if E_root:
        energy_levels.append(E_root / 1.60218e-19)  # Convert Joules back to eV

# Output results
print("First six energy levels (in eV):")
for i, E in enumerate(energy_levels):
    print(f"n = {i+1}: E = {E:.3f} eV")
```
